```python
SEED_NUM = 4   # change if your last digit is different
STUDENT_MAJOR = "BSME"
FAVORITE_ARTIST = "LANA DEL REY"
LOCAL_HAZARD = "FLOOD"

CONTROL_NUM = max(3, SEED_NUM)
```

```python
"Exercise 1"

from collections import defaultdict, Counter

stream = [
    ("TUP Manila", "BSECE"),
    ("TUP Taguig", f"BSME_v{CONTROL_NUM}"),
    ("TUP Manila", f"BSME_v{CONTROL_NUM}"),
    ("TUP Manila", "BSECE"),
    ("TUP Visayas", STUDENT_MAJOR),
    ("TUP Taguig", "BSECE"),
    ("TUP Manila", "BSECE")
]

# Add (CONTROL_NUM + 3) custom entries
for i in range(CONTROL_NUM + 3):
    stream.append(("TUP Manila", STUDENT_MAJOR))

# Aggregation
campus_data = defaultdict(list)
for campus, program in stream:
    campus_data[campus].append(program)

# Frequency analysis
manila_counter = Counter(campus_data["TUP Manila"])
top_program, frequency = manila_counter.most_common(1)[0]

print("CONTROL_NUM:", CONTROL_NUM)
print("STUDENT_MAJOR:", STUDENT_MAJOR)
print("Expanded Stream Length:", len(stream))
print("Total Applications for TUP Manila:", len(campus_data["TUP Manila"]))
print("Top Requested Program at TUP Manila:", top_program)
print("Exact Frequency:", frequency)
```

```python
"Exercise 2"

festival = {"Ben&Ben", "SB19", "Bini", "Eraserheads", FAVORITE_ARTIST, "Zild", f"Indie Artist {CONTROL_NUM}"}

user_a = {"Ben&Ben", "Bini", "Maki", "Dionela", FAVORITE_ARTIST}
user_b = {"SB19", "Eraserheads", "Zild", f"Indie Artist {CONTROL_NUM}", "Parokya ni Edgar"}

# Add CONTROL_NUM new artists
for i in range(CONTROL_NUM):
    festival.add(f"Extra Artist {i}")

# Jaccard calculations
intersection_a = user_a & festival
union_a = user_a | festival
jaccard_a = (len(intersection_a) / len(union_a)) * 100

intersection_b = user_b & festival
union_b = user_b | festival
jaccard_b = (len(intersection_b) / len(union_b)) * 100

# Dealbreakers
dealbreakers = user_a - festival

print("Total Artists in Expanded Festival:", len(festival))
print("Length of Intersection (User A & Festival):", len(intersection_a))
print("Length of Union (User A & Festival):", len(union_a))
print("User A Jaccard Similarity (%):", round(jaccard_a, 2))
print("User B Jaccard Similarity (%):", round(jaccard_b, 2))
print("User A Dealbreaker Artists:", dealbreakers)
```

```python
"Exercise 3"

from collections import deque, Counter

queue = deque(maxlen=5)

burst = [
    (CONTROL_NUM, "WEATHER"),
    (CONTROL_NUM + 1, "TRAFFIC"),
    (CONTROL_NUM + 2, "WEATHER"),
    (CONTROL_NUM + 3, "WEATHER"),
    (CONTROL_NUM + 4, LOCAL_HAZARD),
    (CONTROL_NUM + 5, "WEATHER"),
    (CONTROL_NUM + 6, "TRAFFIC")
]

# Add (CONTROL_NUM + 2) alerts
start = CONTROL_NUM + 7
for i in range(CONTROL_NUM + 2):
    burst.append((start + i, "WEATHER"))

log_weather = "N"
log_hazard = "N"

# Process burst
for timestamp, category in burst:
    categories = [c for _, c in queue]
    count = Counter(categories)

    if count[category] >= 2:
        continue

    queue.append((timestamp, category))

    if timestamp == CONTROL_NUM + 3:
        log_weather = "Y"

    if category == LOCAL_HAZARD:
        log_hazard = "Y"

# Sum timestamps
total_sum = sum(t for t, _ in queue)

print("LOCAL_HAZARD Used:", LOCAL_HAZARD)
print("Total Burst Size:", len(burst))
print("Did WEATHER at CONTROL_NUM+3 append?:", log_weather)
print("Did LOCAL_HAZARD append?:", log_hazard)
print("Final Deque Length:", len(queue))
print("Final Sum of Timestamps:", total_sum)
print("Final Deque Output:", queue)
```
